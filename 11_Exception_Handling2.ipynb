Q75. Create a transaction context manager for banking case 1; use sample reference ACC-1075 and explain the result.

Keep business-rule failures distinct from programming or infrastructure failures.
For medium/difficult work, preserve the original cause and protect partially updated balances.

```python
class Transaction:
    def __init__(self, balance):
        self.balance = balance
        self.original_balance = balance

    def __enter__(self):
        print("Transaction started.")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type:
            self.balance = self.original_balance
            print("Transaction failed. Rolled back.")
        else:
            print("Transaction committed.")

        return False


account = "ACC-1075"
balance = 5000.00

try:
    with Transaction(balance) as transaction:
        transaction.balance -= 1500

        raise ValueError("Transaction processing failed.")

except ValueError as e:
    print("Account:", account)
    print("Error:", e)
```
