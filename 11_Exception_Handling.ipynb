Q33. Process transfers while isolating bad rows for banking case 1; use sample reference ACC-1033 and explain the result.

```python
account = "ACC-1033"

transfers = [
    ("ACC-2001", 1000),
    ("ACC-2002", -500),
    ("ACC-2003", 750)
]

successful = []
failed = []

for target, amount in transfers:
    try:
        if amount <= 0:
            raise ValueError("Transfer amount must be positive.")

        successful.append((target, amount))

    except ValueError as e:
        failed.append((target, str(e)))

print("Account:", account)
print("Successful transfers:", successful)
print("Failed transfers:", failed)
```

Q34. Retry a transient payment gateway for banking case 1; use sample reference ACC-1034 and explain the result.

Q49. Separate validation and recovery for banking case 2; use sample reference ACC-1049 and explain the result.

```python
account = "ACC-1049"
balance = 4000.00
withdrawal = 5000.00

try:
    if withdrawal <= 0:
        raise ValueError("Withdrawal must be positive.")

    if withdrawal > balance:
        raise ValueError("Insufficient balance.")

except ValueError as e:
    print("Validation failed:", e)
    print("Recovery: withdrawal cancelled.")

else:
    balance -= withdrawal
    print("Account:", account)
    print("New balance:", balance)
```
