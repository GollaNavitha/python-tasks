```python
import re

text = 'ACC-1048 deposited ₹4500 and $200'

pattern = re.compile(
    r'(?P<currency>₹|\$)\s*(?P<amount>\d+(?:\.\d+)?)'
)

for match in pattern.finditer(text):
    print('Currency:', match.group('currency'))
    print('Amount:', match.group('amount'))
```

Q50. Find repeated account words for banking case 2; use sample reference ACC-1050 and explain the result.
Hints

Write representative valid and invalid samples before writing the pattern.
Use raw strings, named groups for extraction, and fullmatch for whole-field validation.
Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.
Edge cases to test: empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

text = 'account account transfer balance balance account'

pattern = re.compile(
    r'\b(?P<word>\w+)\s+(?P=word)\b',
    re.IGNORECASE
)

matches = pattern.findall(text)

print('Repeated words:', matches)
```
