```python
class BankAccount:
    def __init__(self, reference, balance):
        self.reference = reference
        self.__balance = balance
        self.__reserved = 0

    @property
    def available_balance(self):
        return self.__balance - self.__reserved

    def reserve(self, amount):
        if amount <= 0:
            raise ValueError("Reservation amount must be positive")
        if amount > self.available_balance:
            raise ValueError("Insufficient available balance")

        self.__reserved += amount

    def release(self, amount):
        if amount <= 0 or amount > self.__reserved:
            raise ValueError("Invalid release amount")
        self.__reserved -= amount


account = BankAccount("ACC-1056", 5000)

account.reserve(1000)

print("Reference:", account.reference)
print("Available Balance:", account.available_balance)
```

```python
class UnitOfWork:
    def __init__(self, balance):
        self.original = balance
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient balance")
        self.balance -= amount
    def rollback(self):
        self.balance = self.original
uow = UnitOfWork(5000)

try:
    uow.withdraw(1000)
    uow.withdraw(5000)
except ValueError as e:
    uow.rollback()
    print("Transaction rolled back:", e)

print("Final balance:", uow.balance)
```

```python
class UnitOfWork:
    def __init__(self, balance):
        self.original_balance = balance
        self.balance = balance
    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient balance")
        self.balance -= amount
    def rollback(self):
        self.balance = self.original_balance
uow = UnitOfWork(5000)
try:
    uow.withdraw(1000)
    uow.withdraw(5000)
except ValueError as e:
    uow.rollback()
    print("Transaction rolled back:", e)

print("Final balance:", uow.balance)
```
