Edge cases to test: empty input, invalid types, boundary values, and a realistic failure path.

```python
def log_checkout(func):
    def wrapper(amount):
        print("Checkout called")
        return func(amount)
    return wrapper

@log_checkout
def checkout(amount):
    return amount

print(checkout(8000))
```
