```python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees
# Rows represent stores
# Columns represent product categories

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees
# Rows represent stores
# Columns represent product categories

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
```

Question 45: Premium Performance A store-category combination is considered premium when:

Sales are greater than ₹5,00,000 Profit is greater than ₹1,50,000 Find all premium-performing combinations.

```python
profit=sales_data - cost_data
premium=np.logical_and(sales_data > 500000,profit > 150000)
store_index,category_index=np.where(premium)
print(sales_data[store_index,category_index])
print(profit[store_index,category_index])
```

Question 46: Underperforming Combinations A store-category combination is considered underperforming when:

Sales are below ₹2,50,000 Profit margin is below 25% Find all such combinations.

```python
profit=sales_data-cost_data
profit_margin=(profit/sales_data)*100
underperforming=np.logical_and(sales_data < 250000,profit_margin < 25)
store_index,category_index=np.where(underperforming)
for i,j in zip(store_index,category_index):
  print("store:",stores[i])
  print("category:",categories[j])
  print("profit:",profit[i,j])
  print("sales:",sales_data[i,j])
  print("profit_margin:",profit_margin[i,j])
```

Question 47: Category Performance Classification Create a new array that classifies every sales value as:

"High" when sales are at least ₹5,00,000 "Medium" when sales are between ₹2,50,000 and ₹4,99,999 "Low" when sales are below ₹2,50,000 Use np.where() or np.select().

```python
classification=np.where(
    sales_data >= 500000,
    "high",np.where((sales_data >= 250000) & (sales_data <499999),"medium","low"))

print(classification)
```
