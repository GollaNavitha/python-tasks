```python
#2 Alphabetical order if lengths are the same
def compare_words(word1, word2):
    if len(word1) > len(word2):
        return word1
    elif len(word2) > len(word1):
        return word2
    else:
        if word1 < word2:
            return word1
        else:
            return word2

word1 = "Mango"
word2 ="Banana"

result = compare_words(word1, word2)
print(result)
```

```python
#Q10: Simple Interest Calculator
def simple_interest(principal, rate, time):
    return (principal * rate * time) / 100

principal = float(input("Enter principal amount: "))
rate = float(input("Enter rate of interest: "))
time = float(input("Enter time (in years): "))

interest = simple_interest(principal, rate, time)
print("Simple Interest:", interest)
```

```python
#Q12: Check Leap Year
def check_leap_year(year):
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        return "Leap Year"
    else:
        return "Not a Leap Year"

year = int(input("Enter a year: "))

result = check_leap_year(year)
print(result)
```

```python
#Q15: Count Vowels in a String.
def count_vowels(text):
    count = 0
    for ch in text:
        if ch in "aeiouAEIOU":
            count += 1
    return count
string = input("Enter a string: ")
result = count_vowels(string)
print("Number of vowels:", result)
```

```python
#Q34: Prime Number Function
def is_prime(num):
    if num <= 1:
        return "Not a Prime Number"
    for i in range(2, num):
        if num % i == 0:
            return "Not a Prime Number"
    return "Prime Number"
number = int(input("Enter a number: "))
result = is_prime(number)
print(result)
```

```python
#Q88: Three-Level Nested Functions
def outer():
    print("Outer function")
    def middle():
        print("Middle function")
        def inner():
            print("Inner function")
        inner()
    middle()
outer()
```

```python
#Q92: Closure-Based Bank Balance
def bank_balance(balance):
    def deposit(amount):
        return balance + amount
    return deposit
account = bank_balance(1000)
print(account(500))
```

```python
 #Q96: Modify a Global Variable
count = 10
def increase():
    global count
    count = count + 5
increase()
print("Count =", count)
```

```python
#Q98: Modify an Enclosing Variable Using nonlocal
def outer():
    count = 10
    def inner():
        nonlocal count
        count = count + 5
    inner()
    print("Count =", count)
outer()
```

```python
 #Q99: LEGB Name Resolution
name = "Global"
def outer():
    name = "Enclosing"
    def inner():
        name = "Local"
        print(name)
    inner()
outer()
print(name)
```
