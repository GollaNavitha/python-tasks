```python
class Account:
    def __init__(self, account_number, amount):
        self.account_number = account_number
        self.amount = amount

    @property
    def amount(self):
        return self._amount

    @amount.setter
    def amount(self, value):
        if value < 0:
            raise ValueError("Amount cannot be negative.")
        self._amount = value


account = Account("ACC-1008", 5000)

print("Account:", account.account_number)
print("Amount:", account.amount)

account.amount = 6000

print("Updated amount:", account.amount)
```

```python
class Account:
    def __init__(self, account_ref, balance):
        self.account_ref = account_ref
        self.__balance = balance
        self.__reserved = 0

    def reserve_balance(self, amount):
        if amount <= 0:
            print("Reservation amount must be positive.")
        elif amount > self.__balance - self.__reserved:
            print("Insufficient available balance.")
        else:
            self.__reserved += amount
            print("Amount reserved successfully.")

    def available_balance(self):
        return self.__balance - self.__reserved


account = Account("ACC-1041", 10000)

account.reserve_balance(3000)

print("Account:", account.account_ref)
print("Available Balance:", account.available_balance())
```

Q44. Implement a repository abstraction for banking case 1; use sample reference ACC-1044 and explain the result.
Hints

List the objects, their responsibilities, and their invariants before coding.
Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
Keep balances private and expose controlled deposit/withdraw/transfer behavior.
Edge cases to test: empty input, invalid types, boundary values, and a realistic failure path.

```python
from abc import ABC, abstractmethod


class AccountRepository(ABC):

    @abstractmethod
    def save(self, account):
        pass

    @abstractmethod
    def find(self, reference):
        pass


class MemoryAccountRepository(AccountRepository):

    def __init__(self):
        self.accounts = {}

    def save(self, account):
        self.accounts[account["reference"]] = account

    def find(self, reference):
        return self.accounts.get(reference)


repository = MemoryAccountRepository()

account = {
    "reference": "ACC-1044",
    "balance": 25000
}

repository.save(account)

print(repository.find("ACC-1044"))
```
