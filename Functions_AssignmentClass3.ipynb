```python
## 1.Using filter() and map():

#Extract all even numbers.
#Calculate the square of every even number.
#Return the squared values as a list.
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
def is_even(num):
    return num % 2 == 0

def square(num):
    return num ** 2

result = list(map(square, filter(is_even, numbers)))
print(result)
```

```python
## 2.Using filter() and reduce():

#Extract all odd numbers.
#Calculate their total sum.
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
def is_odd(num):
    return num % 2 != 0

def add(x, y):
    return x + y

result = reduce(add, filter(is_odd, numbers))
print(result)
```

```python
## 3.Using filter() and map():

#Extract numbers greater than 50.
#Multiply every selected number by 2.

numbers = [10, 25, 40, 55, 70, 85, 100]
def greater_than_50(num):
    return num > 50

def multiply_by_2(num):
    return num * 2

result = list(map(multiply_by_2, filter(greater_than_50, numbers)))
print(result)
```

```python
## 4.Using filter(), map(), and reduce():

#Extract positive numbers.
#Calculate the square of every positive number.
#Calculate the total sum of the squared values.

numbers = [-15, 20, -35, 40, 50, -10, 60]

def is_positive(num):
    return num > 0

def square(num):
    return num ** 2

def add(x, y):
    return x + y

result = reduce(add, map(square, filter(is_positive, numbers)))
print(result)
```

```python
## 5.Using filter() and map():

#Extract words containing more than five characters.
#Convert the selected words to uppercase.

words = ["python", "azure", "ai", "developer", "cloud", "analytics", "spark"]
def more_than_five_characters(word):
    return len(word) > 5

def to_uppercase(word):
    return word.upper()

result = list(map(to_uppercase, filter(more_than_five_characters, words)))
print(result)
     
```

```python
## 6.Using filter() and map():

#Extract names beginning with "A".
#Convert the selected names to lowercase.

names = ["Anil", "Rahul", "Ajay", "Meena", "Arjun", "Kiran", "Akhil"]
def starts_with_a(name):
    return name.startswith("A")

def to_lowercase(name):
    return name.lower()

result = list(map(to_lowercase, filter(starts_with_a, names)))
print(result)
```

```python
## 7.Using filter() and reduce():

#Extract numbers divisible by both 3 and 5.
#Calculate their total sum.

numbers = [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]
def divisible_by_3_and_5(num):
    return num % 3 == 0 and num % 5 == 0

def add(x, y):
    return x + y

result = reduce(add, filter(divisible_by_3_and_5, numbers))
print(result)
     
```

```python
## 8.Using filter() and reduce():

#Extract palindrome words.
#Join them into one string separated by " | ".

words = ["madam", "python", "level", "azure", "radar", "cloud", "civic"]
def is_palindrome(word):
    return word == word[::-1]

def join_with_pipe(x, y):
    return x + " | " + y

result = reduce(join_with_pipe, filter(is_palindrome, words))
print(result)
```

```python
## 9.Using filter(), map(), and reduce():

#Extract temperatures greater than 30°C.
#Convert the selected values into Fahrenheit.
#Calculate the average Fahrenheit temperature.

temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]
def greater_than_30(temp):
    return temp > 30

def to_fahrenheit(temp):
    return (temp * 9/5) + 32

def add(x, y):
    return x + y

result = reduce(add, map(to_fahrenheit, filter(greater_than_30, temperatures))) / len(list(filter(greater_than_30, temperatures)))
print(result)
```

```python
## 10.Using filter(), map(), and reduce():

#Extract prices greater than 1000.
#Apply a 10% discount to every selected price.
#Calculate the total discounted amount.

prices = [450, 1200, 2500, 800, 5000, 1500, 7500]
def greater_than_1000(price):
    return price > 1000

def apply_discount(price):
    return price * 0.9

def add(x, y):
    return x + y

result = reduce(add, map(apply_discount, filter(greater_than_1000, prices)))
print(result)
     
```

```python
## 11.Using filter() and map():

#Extract employees earning more than 60000.
#Increase their salary by 10%.
#Return their names and updated salaries.
employees = [
    {"id": 101, "name": "Rahul", "salary": 45000},
    {"id": 102, "name": "Anil", "salary": 65000},
    {"id": 103, "name": "Sneha", "salary": 72000},
    {"id": 104, "name": "Kiran", "salary": 82000},
    {"id": 105, "name": "Meena", "salary": 55000}
]
def high_salary(employee):
    return employee["salary"] > 60000

def increase_salary(employee):
    return {
        "name": employee["name"],
        "salary": employee["salary"] * 1.10
    }

result = list(map(increase_salary, filter(high_salary, employees)))
print(result)
```

```python
## 13.Using filter(), map(), and reduce():

#Extract active employees.
#Increase every active employee's salary by 15%.
#Calculate the total updated payroll.

employees = [
    {"id": 101, "name": "Rahul", "salary": 45000, "active": True},
    {"id": 102, "name": "Anil", "salary": 65000, "active": True},
    {"id": 103, "name": "Sneha", "salary": 72000, "active": False},
    {"id": 104, "name": "Kiran", "salary": 82000, "active": True},
    {"id": 105, "name": "Meena", "salary": 55000, "active": True},
    {"id": 106, "name": "Arjun", "salary": 98000, "active": False}
]

def is_active(employee):
    return employee["active"]

def increase_salary(employee):
    return {
        "id": employee["id"],
        "name": employee["name"],
        "salary": employee["salary"] * 1.15,
        "active": employee["active"]
    }

def total_payroll(total, employee):
    return total + employee["salary"]

result = reduce(
    total_payroll,
    map(increase_salary, filter(is_active, employees)),
    0
)

print(result)
```

```python
## 14.Using filter() and reduce():

#Extract students who passed.
#Find the topper among the passed students.

students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]
def passed_student(student):
    return student["passed"]

def topper(student1, student2):
    return student1 if student1["marks"] > student2["marks"] else student2

result = reduce(topper, filter(passed_student, students))

print(result)
     
```

```python
## 15.Using filter(), map(), and reduce():

#Extract students scoring more than 70.
#Extract only their marks.
#Calculate the average marks.

students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]
def above_70(student):
    return student["marks"] > 70

def get_marks(student):
    return student["marks"]

def add(total, mark):
    return total + mark

selected_marks = list(map(get_marks, filter(above_70, students)))

total = reduce(add, selected_marks, 0)
average = total / len(selected_marks)

print("Average Marks:", average)
     
```

```python
## 16.Using filter(), map(), and reduce():

#Extract products currently in stock.
#Calculate each product's inventory value using price × stock.
#Calculate the total inventory value.

products = [

    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]
def in_stock(product):
    return product["stock"] > 0

def inventory_value(product):
    return product["price"] * product["stock"]

def total_value(total, value):
    return total + value

values = list(map(inventory_value, filter(in_stock, products)))

result = reduce(total_value, values, 0)

print("Total Inventory Value:", result)
```

```python
## 17.Using filter(), map(), and reduce():

#Extract products belonging to the Electronics category.
#Apply a 20% discount.
#Calculate the total discounted value.

products = [
    {"name": "Laptop", "price": 55000, "category": "Electronics"},
    {"name": "Chair", "price": 4500, "category": "Furniture"},
    {"name": "Monitor", "price": 18000, "category": "Electronics"},
    {"name": "Desk", "price": 8500, "category": "Furniture"},
    {"name": "SSD", "price": 6500, "category": "Electronics"}
]
def is_electronics(product):
    return product["category"] == "Electronics"

def apply_discount(product):
    return product["price"] * 0.80   # 20% discount

def total_price(total, price):
    return total + price

discounted_prices = list(map(apply_discount, filter(is_electronics, products)))

result = reduce(total_price, discounted_prices, 0)

print("Total Discounted Value:", result)
```

```python
## 19.Using filter(), map(), and reduce():

#Extract transactions whose amount is greater than 5000.
#Add 18% GST to every selected transaction.
#Calculate the grand total after GST.
transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]
def greater_than_5000(transaction):
    return transaction["amount"] > 5000

def add_gst(transaction):
    return transaction["amount"] * 1.18

def grand_total(total, amount):
    return total + amount

gst_amounts = list(map(add_gst, filter(greater_than_5000, transactions)))

result = reduce(grand_total, gst_amounts, 0)

print("Grand Total After GST:", result)
```

```python
## 20.Using filter(), map(), and reduce():

#Extract active salespeople from the South region.
#Calculate a 5% incentive for each selected salesperson.
#Create a list containing salesperson name, sales amount, and incentive.
#Calculate the total incentive payable.
sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]
def is_active_south(record):
    return record["region"] == "South" and record["active"]

def calculate_incentive(record):
    return {
        "salesperson": record["salesperson"],
        "sales": record["sales"],
        "incentive": record["sales"] * 0.05
    }

def total_incentive(total, record):
    return total + record["incentive"]

selected = list(map(calculate_incentive,
                    filter(is_active_south, sales_records)))

print("Selected Salespeople:")
print(selected)

result = reduce(total_incentive, selected, 0)
print("Total Incentive Payable:", result)
```
