Q98. Implement retry as a higher-transfer function for banking case 3; use sample reference ACC-1098 and explain the result.

Test the pipeline with an empty collection and avoid mutating account records in place.
Edge cases to test: empty input, invalid types, boundary values, and a realistic failure path.

```python
def retry(func, attempts):
    for _ in range(attempts):
        try:
            return func()
        except ValueError:
            pass
    return "Transfer failed"

count = 0

def transfer():
    global count
    count += 1
    if count < 2:
        raise ValueError("Temporary failure")
    return "Transfer successful"

print(retry(transfer, 3))
```

```python
from functools import reduce

transactions = [3000, -500, 2000, 1000, -200]

credits = filter(lambda x: x > 0, transactions)
amounts = map(lambda x: x * 1.05, credits)
total = reduce(lambda a, b: a + b, amounts, 0)

print(round(total, 2))
```
